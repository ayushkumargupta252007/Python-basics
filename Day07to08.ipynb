```python
import matplotlib.pyplot as plt
import numpy as np
data = np.array([
[90, 80, 70],
[60, 75, 85],
[95, 88, 92],
[70, 65, 78]
])
students = ["Amit", "Rahul", "Priya", "Neha"]
subjects = ["Math", "Science", "English"]
plt.imshow(data)
plt.xticks(range(len(subjects)), subjects)
plt.yticks(range(len(students)), students)
plt.colorbar()
plt.title("Student Marks Heat Map")
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
sales = np.array([
[25, 30, 28],
[35, 40, 38],
[20, 18, 22],
[45, 50, 48]
])
months = ["Jan", "Feb", "Mar", "Apr"]
products = ["Laptop", "Mobile", "Tablet"]
plt.imshow(sales, cmap="Oranges")
plt.xticks(range(len(products)), products)
plt.yticks(range(len(months)), months)
plt.colorbar(label="Sales")
plt.title("Monthly Sales Heat Map")
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
temperature = np.array([
[30, 32, 31],
[35, 36, 34],
[28, 29, 30],
[40, 39, 41]
])
cities = ["Delhi", "Mumbai", "Shimla", "Jaipur"]
days = ["Mon", "Tue", "Wed"]
plt.imshow(temperature, cmap="coolwarm")
plt.xticks(range(len(days)), days)
plt.yticks(range(len(cities)), cities)
plt.colorbar(label="Temperature (°C)")
plt.title("Temperature Heat Map")
plt.show()
```
