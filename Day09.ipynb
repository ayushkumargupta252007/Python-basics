```python
import pandas as pd
from sklearn.tree import DecisionTreeClassifier, plot_tree
from sklearn.preprocessing import LabelEncoder
import matplotlib.pyplot as plt

# Build the dataset matching your counts
data = (
    [{'purpose': 'gaming', 'brand': 'Asus'}] * 10 +
    [{'purpose': 'gaming', 'brand': 'Dell'}] * 5 +
    [{'purpose': 'study',  'brand': 'Dell'}] * 15 +
    [{'purpose': 'study',  'brand': 'HP'}]   * 5 +
    [{'purpose': 'office', 'brand': 'Dell'}] * 10
)
df = pd.DataFrame(data)

# Encode purpose as numbers (Dell/HP/Asus target stays as labels for readability)
le = LabelEncoder()
df['purpose_encoded'] = le.fit_transform(df['purpose'])

X = df[['purpose_encoded']]
y = df['brand']

# Train the tree
model = DecisionTreeClassifier(criterion='gini', random_state=42)
model.fit(X, y)

# Check accuracy on the same data (small dataset, just for sanity check)
print("Training accuracy:", model.score(X, y))

# Visualize
plt.figure(figsize=(10, 6))
plot_tree(
    model,
    feature_names=['purpose'],
    class_names=model.classes_,
    filled=True,
    rounded=True
)
plt.show()

# Predict brand for a new purpose
sample = pd.DataFrame({'purpose_encoded': le.transform(['gaming'])})
print("Predicted brand for gaming:", model.predict(sample)[0])
```

```python
from sklearn.linear_model import LinearRegression
# House size
X = [[1000], [1200], [1500], [1800]]
# House price (in lakh)
y = [40, 50, 65, 80]
model = LinearRegression()
model.fit(X, y)
prediction = model.predict([[1600]])
print("Predicted Price:", prediction[0], "Lakh")
```

```python
from sklearn.cluster import KMeans
import numpy as np
data = np.array([
[2, 3],
[3, 4],
[10, 12],
[11, 13]
])
model = KMeans(n_clusters=2, random_state=42)
model.fit(data)
print("Cluster Labels:")
print(model.labels_)
```

```python
from sklearn.linear_model import LinearRegression

experience = [[1], [2], [3], [4], [5]]
salary = [25000, 35000, 45000, 55000, 65000]
model = LinearRegression()
model.fit(experience, salary)
prediction = model.predict([[6]])

print("Predicted Salary =", prediction[0])
print("Coefficient (slope):", model.coef_[0])
print("Intercept:", model.intercept_)
```
